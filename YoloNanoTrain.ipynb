```python
!pip install ultralytics
from ultralytics import YOLO
from google.colab import drive
import shutil

# 1. חיבור ל-Google Drive
drive.mount('/content/drive')

# 2. Load yolo 26 nano to start training on our data
#model = YOLO('yolo26n.pt')
model = YOLO('yolo26s.pt')

# 3. הגדרת נתיב לקובץ התצורה שהכנו
# שים לב שזהו הנתיב בו שמרנו את ה-data.yaml בסקריפט הקודם
data_path = '/content/drive/My Drive/DeepLearning/unified_dataset/data.yaml'

# 4. הרצת תהליך האימון (Training)
print("מתחיל באימון הרשת...")
results = model.train(
    data=data_path,
    epochs=300,                 # מספר הפעמים שהרשת תעבור על כל ה-Dataset
    imgsz=640,                  # גודל התמונה שהרשת "רואה" (640x640 זה סטנדרט מעולה ל-YOLO)
    batch=16,                   # גודל האצווה (כמה תמונות הרשת מעבדת יחד בכל צעד). אם יש שגיאת זיכרון, שנה ל-8.
    name='pest_detection_yolo26s', # השם של תיקיית התוצאות שתווצר
    device=0,                   # פקודה להשתמש בכרטיס המסך (GPU)
    patience=30                 # מנגנון Early Stopping: אם המודל לא משתפר במשך 30 אפוקים רצופים, האימון יעצור כדי למנוע Overfitting
)

print("האימון הסתיים בהצלחה!")

# 5. הערכת ביצועים (Validation)
# המודל ירוץ על תיקיית ה-valid ויבדוק כמה הוא באמת מדויק
metrics = model.val()
print(f"mAP50-95 (דיוק ממוצע כולל): {metrics.box.map:.3f}")
print(f"mAP50 (דיוק בחפיפה של 50%): {metrics.box.map50:.3f}")

# העתקת התוצאות לתיקייה בדרייב שלי
source_dir = '/content/runs/detect/pest_detection_nano'
destination_dir = '/content/drive/My Drive/DeepLearning/TrainedModel/pest_detection_nano_yolo26s_epoch300'
try:
    # השתמש ב-shutil.copytree כדי להעתיק את כל התיקייה
    shutil.copytree(source_dir, destination_dir)
    print(f"התוצאות הועתקו בהצלחה מ-'{source_dir}' ל-'{destination_dir}'")
except FileExistsError:
    print(f"התיקייה '{destination_dir}' כבר קיימת. בבקשה מחק אותה ידנית לפני ההעתקה מחדש, או שנה את שם היעד.")
except Exception as e:
    print(f"אירעה שגיאה בעת העתקת הקבצים: {e}")
```
