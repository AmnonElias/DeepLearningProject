```python
import ultralytics
from ultralytics import YOLO
from google.colab import drive

# 1. חיבור ל-Google Drive
drive.mount('/content/drive')

# 2. Load yolo 12 nano to start training on our data
model = YOLO('yolov8n.pt')

# 3. הגדרת נתיב לקובץ התצורה שהכנו
# שים לב שזהו הנתיב בו שמרנו את ה-data.yaml בסקריפט הקודם
data_path = '/content/drive/My Drive/DeepLearning/unified_dataset/data.yaml'

# 4. הרצת תהליך האימון (Training)
print("מתחיל באימון הרשת...")
results = model.train(
    data=data_path,
    epochs=50,                  # מספר הפעמים שהרשת תעבור על כל ה-Dataset (כדאי להתחיל ב-50 ולעלות ל-100-300 בהמשך)
    imgsz=640,                  # גודל התמונה שהרשת "רואה" (640x640 זה סטנדרט מעולה ל-YOLO)
    batch=16,                   # גודל האצווה (כמה תמונות הרשת מעבדת יחד בכל צעד). אם יש שגיאת זיכרון, שנה ל-8.
    name='pest_detection_nano', # השם של תיקיית התוצאות שתווצר
    device=0,                   # פקודה להשתמש בכרטיס המסך (GPU)
    patience=15                 # מנגנון Early Stopping: אם המודל לא משתפר במשך 15 אפוקים רצופים, האימון יעצור כדי למנוע Overfitting
)

print("האימון הסתיים בהצלחה!")

# 5. הערכת ביצועים (Validation)
# המודל ירוץ על תיקיית ה-valid ויבדוק כמה הוא באמת מדויק
metrics = model.val()
print(f"mAP50-95 (דיוק ממוצע כולל): {metrics.box.map:.3f}")
print(f"mAP50 (דיוק בחפיפה של 50%): {metrics.box.map50:.3f}")
```
